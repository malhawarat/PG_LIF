# PG-LIF: method overview figure
Generates `fig0_overview.png` (300 dpi), the schematic of Figure 1. Symbols and equation numbers follow
Sections 3 and 5 of the manuscript.

```python
import matplotlib
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch
plt.rcParams.update({'font.family': 'DejaVu Sans', 'font.size': 10, 'mathtext.fontset': 'dejavusans'})
fig, ax = plt.subplots(figsize=(11.5, 6.2)); ax.set_xlim(0, 117); ax.set_ylim(-4, 64); ax.axis('off')
EDGE = '#4A4A4A'; LS = 4.0
def box(x, y, w, title, lines, fc):
    h = 6.5 + LS * len(lines)
    ax.add_patch(FancyBboxPatch((x, y), w, h, boxstyle='round,pad=0.5,rounding_size=1.2', fc=fc, ec=EDGE, lw=1.1))
    ax.text(x + w/2, y + h - 1.6, title, ha='center', va='top', fontsize=10.5, fontweight='bold')
    for i, l in enumerate(lines): ax.text(x + w/2, y + h - 6.0 - LS*i, l, ha='center', va='top', fontsize=9.2)
    return h
def arrow(p, q, text=None, tpos=None, color=EDGE, ls='-', rad=0.0, fs=8.8, bg=False):
    ax.add_patch(FancyArrowPatch(p, q, arrowstyle='-|>', mutation_scale=13, lw=1.3, color=color, ls=ls, connectionstyle=f'arc3,rad={rad}'))
    if text: ax.text(*tpos, text, ha='center', va='bottom', fontsize=fs, color=color,
                     bbox=dict(fc='white', ec='none', pad=1.0) if bg else None)
box(1, 24, 15, ['spikes $x(t)$', r'current $W_{\mathrm{in}}x(t)$'], '#E8EEF7') if False else None
hin = box(1, 23, 16, 'Input', [r'spikes $x(t)$', r'current $W_{\mathrm{in}}x(t)$'], '#E8EEF7')
hd = box(24, 36, 21, 'Dendrite', [r'$V_d$, leak $\tau_d$ (Eq. 1)', r'event when $V_d \geq \theta_d$', r'outside refractory $\tau_{\mathrm{ref},p}$', r'event times $t_{k,d}$ (Eq. 2)'], '#DCEBDD')
hp = box(52, 36, 21, 'Plateau state', [r'$p$ (Eq. 3)', r'jumps by $P_0$ at each $t_{k,d}$', r'decays with $\tau_p$', r'bounded (Prop. 2)'], '#FBE5C8')
hs = box(24, 3, 21, 'Soma', [r'$V_s$, leak $\tau_m$ (Eq. 4)', r'spike when $V_s \geq \theta_0+\beta a$', r'subtractive reset', r'adaptation $a$ (Eq. 5)'], '#E3DDF0')
hl = box(85, 32, 31, 'Plasticity (three-factor)', [r'gate $g_i=\min(1,\,p_i/p_{\max})$ (Eq. 13)', r'eligibility trace $e_{ij}$', r'broadcast learning signal $\ell$', r'$\Delta w_{ij}=\eta\,\ell\,g_i\,e_{ij}$ (Eq. 14)', r'one-shot variant (Eq. 15)'], '#F6D9D9')
ho = box(85, 7, 31, 'Output', [r'spikes $s(t)$ at times $t_k$', r'to next layer, recurrence, readout'], '#EFEFEF')
yin = 23 + hin/2
arrow((17.6, yin + 2), (23.4, 36 + hd/2), None); ax.text(17.6, 41, r'$I_d$', fontsize=10, color=EDGE)
arrow((17.6, yin - 2), (23.4, 3 + hs/2), None); ax.text(17.6, 13.5, r'$I_s$', fontsize=10, color=EDGE)
arrow((45.6, 36 + hd/2), (51.4, 36 + hp/2), r'$e_d(t)$', (48.5, 36 + hd/2 + 0.8))
arrow((62.5, 35.4), (45.6, 3 + hs*0.7), r'role 1: somatic drive $\kappa\,p$', (46.5, 27.6), color='#B5651D', rad=-0.12)
arrow((73.6, 36 + hp/2), (84.4, 36 + hp/2), r'role 2: gate', (79.0, 36 + hp/2 + 1.0), color='#B03A2E', fs=8.0)
arrow((45.6, 3 + hs/2), (84.4, 7 + ho/2), r'$s(t)$', (70, 3 + hs/2 + 0.6))
arrow((58, 3 + hs/2), (97, 31.4), r'surrogate $\psi_s \to e_{ij}$', (73, 21.5), color='#666666', ls='--', rad=-0.12, fs=8.4, bg=True)
ax.text(1, 62, 'Forward dynamics (Section 3)', fontsize=10.5, color='#333333', fontweight='bold')
ax.text(85, 62, 'Local learning (Section 5)', fontsize=10.5, color='#333333', fontweight='bold')
ax.text(1, -3.2, 'Regimes A and B: dendrite and soma receive the same feedforward current; the soma also receives recurrent input.', fontsize=8.6, color='#333333')
plt.savefig('fig0_overview.png', dpi=300, bbox_inches='tight', facecolor='white'); plt.show()
```
